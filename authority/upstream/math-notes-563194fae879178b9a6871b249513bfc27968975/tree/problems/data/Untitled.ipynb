```python
import numpy as np
import matplotlib.pyplot as plt

def function_g(x):
    return x**2 / (np.sqrt(x**2 + 1) + 1)

def function_f(x):
    return np.sqrt(x**2 + 1) - 1

x_range_wide = np.single(np.linspace(-1, 1, 101))
x_range_narrow = np.single(np.linspace(-0.002, 0.002, 101))

f_values_wide = function_f(x_range_wide)
g_values_wide = function_g(x_range_wide)
f_values_narrow = function_f(x_range_narrow)
g_values_narrow = function_g(x_range_narrow)

plt.plot(x_range_wide, f_values_wide, color='blue', label='f(x)')
plt.plot(x_range_wide, g_values_wide, color='red', label='g(x)')
plt.title('On Range [-1, 1]')
plt.legend()
plt.show()

plt.plot(x_range_narrow, f_values_narrow, color='blue', label='f(x)')
plt.plot(x_range_narrow, g_values_narrow, color='red', label='g(x)')
plt.title('On Range [-0.002, 0.002]')
plt.legend()
plt.show()
```
